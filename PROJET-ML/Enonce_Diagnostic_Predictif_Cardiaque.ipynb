# ÉNONCÉ DE L’EXERCICE

# Diagnostic Prédictif des Insuffisances Cardiaques

## Objectif du projet

L'objectif de cet exercice est de développer un système d'apprentissage automatique capable de **prédire si un patient présente un risque élevé de maladie cardiaque** (variable cible binaire `HeartDisease` : 0 = Sain, 1 = Malade) à partir de plusieurs indicateurs cliniques et physiologiques.

> **Note pédagogique :** le jeu de données concerne la classification de maladies cardiaques ; il ne constitue pas un diagnostic médical et ne permet pas à lui seul de diagnostiquer une insuffisance cardiaque.

## Présentation du jeu de données

Le fichier `heart.csv` contient **918 observations (patients)** et **11 variables explicatives**. Téléchargez-le depuis Kaggle : [**Heart Failure Prediction Dataset**](https://www.kaggle.com/datasets/fedesoriano/heart-failure-prediction).

### Dictionnaire des variables

| Variable | Description |
|---|---|
| `Age` | Âge du patient (années) |
| `Sex` | Sexe : M = Masculin, F = Féminin |
| `ChestPainType` | Type de douleur thoracique : TA, ATA, NAP, ASY |
| `RestingBP` | Pression artérielle au repos (mm Hg) |
| `Cholesterol` | Cholestérol sérique (mg/dl) |
| `FastingBS` | Glycémie à jeun : 1 si > 120 mg/dl, 0 sinon |
| `RestingECG` | Électrocardiogramme au repos : Normal, ST, LVH |
| `MaxHR` | Fréquence cardiaque maximale atteinte |
| `ExerciseAngina` | Angine induite par l'exercice : Y = Oui, N = Non |
| `Oldpeak` | Dépression du segment ST à l'effort par rapport au repos |
| `ST_Slope` | Pente du segment ST à l'effort : Up, Flat, Down |
| **`HeartDisease` (cible)** | **1 = Maladie cardiaque, 0 = Normal** |

## Travail à réaliser — Les 9 étapes obligatoires

Vous devez rédiger un **script Python structuré** respectant la feuille de route suivante.

### 1. Importation des librairies

- Importer `pandas`, `numpy` et `matplotlib`.
- Importer les outils de prétraitement `sklearn.preprocessing` et `sklearn.compose`.
- Importer les outils de modélisation, d'évaluation et `joblib` pour l'export.

### 2. Chargement des données

- Charger `heart.csv` avec le délimiteur approprié afin d'obtenir un DataFrame exploitable.

### 3. Analyse exploratoire des données (EDA)

- Afficher les dimensions du dataset et ses premières lignes.
- Lister les types de données de chaque colonne.
- Vérifier les valeurs manquantes (`NaN`).
- Calculer la proportion exacte de patients sains et malades.

### 4. Prétraitement des données (*Data Preprocessing*)

- Séparer les variables explicatives (`X`) de la cible (`y`).
- Identifier les variables numériques et catégorielles.
- Encoder les variables textuelles (par exemple `OneHotEncoder`).
- Normaliser les variables numériques (par exemple `StandardScaler`).
- Utiliser de préférence `ColumnTransformer`.

### 5. Division en Train/Test

- Créer un ensemble d'entraînement (**80 %**) et un ensemble de test (**20 %**).
- Utiliser la stratification selon `HeartDisease`.
- Éviter toute fuite de données (*data leakage*) : ajuster les transformations uniquement sur l'entraînement, idéalement via un `Pipeline`.

### 6. Modélisation

Entraîner et comparer **trois modèles de classification** :

1. Réseau de neurones artificiels : `MLPClassifier`, avec historique de perte.
2. Modèle d'ensemble : `RandomForestClassifier`.
3. Modèle linéaire de référence : `LogisticRegression`.

### 7. Observation de la courbe d'entraînement

- Extraire l'historique `loss_curve_` du `MLPClassifier` entraîné.
- Tracer la perte en fonction des itérations/époques afin d'observer la convergence.

### 8. Évaluation des modèles

- Calculer l'**Accuracy** de chacun des trois modèles sur le test.
- Sélectionner le meilleur modèle selon cette métrique.
- Afficher son `classification_report` (précision, rappel et F1-score).
- Visualiser sa matrice de confusion et commenter les faux positifs et faux négatifs.

### 9. Export des modèles

- Sauvegarder le meilleur modèle avec `joblib`.
- Sauvegarder le prétraitement associé, ou exporter directement le `Pipeline` complet, afin de traiter de nouveaux patients de manière identique.
